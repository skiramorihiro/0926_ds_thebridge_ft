## Ejercicio 19
Crea una matriz de 10x1, toda ella compuesta por `False`. Después transformalo en un 2x5

````python
import numpy as np

matriz_inicial = np.full((10, 1), False, dtype=bool)

print("Matriz inicial (10x1):")
print(matriz_inicial)
print("Forma:", matriz_inicial.shape)


matriz_transformada = matriz_inicial.reshape(2, 5)
print("\nMatriz transformada (2x5):")
print(matriz_transformada)
print("Forma:", matriz_transformada.shape)

````
